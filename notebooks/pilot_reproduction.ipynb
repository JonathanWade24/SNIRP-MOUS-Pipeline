```python
artifact_dir = Path("MOUS_A2002_pilot(2)")
if artifact_dir.exists():
    z_ref = np.load(artifact_dir / "sub-A2002_dirs_zinnen.npy")
    w_ref = np.load(artifact_dir / "sub-A2002_dirs_woorden.npy")
    r_ref = np.load(artifact_dir / "sub-A2002_dirs_rest.npy")

    z_new = np.load(Path(cfg.derivatives_root) / "A2002" / "m9_orchestration" / "sub-A2002_dirs_zinnen.npy")
    w_new = np.load(Path(cfg.derivatives_root) / "A2002" / "m9_orchestration" / "sub-A2002_dirs_woorden.npy")
    r_new = np.load(Path(cfg.derivatives_root) / "A2002" / "m9_orchestration" / "sub-A2002_dirs_rest.npy")

    print("zinnen allclose:", np.allclose(z_ref, z_new, atol=1e-6))
    print("woorden allclose:", np.allclose(w_ref, w_new, atol=1e-6))
    print("rest allclose:", np.allclose(r_ref, r_new, atol=1e-6))
else:
    print("Reference artifacts not found; parity check skipped.")
```
